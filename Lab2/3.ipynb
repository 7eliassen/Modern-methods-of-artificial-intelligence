# Задание 3. Преобразование слов в их базовые формы с помощью лемматизации

Используется `WordNetLemmatizer` из NLTK в двух режимах: существительное (`pos='n'`) и глагол (`pos='v'`).

```python
import nltk
from nltk.stem import WordNetLemmatizer

# Словарь WordNet для лемматизатора
for resource in ["wordnet", "omw-1.4"]:
    nltk.download(resource, quiet=True)
```

```python
words = ["sing", "sang", "singer", "singing", "calves", "cats", "running", "runs",
         "ran", "happily", "happiness", "generously", "argue", "argued", "university",
         "universal", "fairly", "arrival", "gently", "maximum"]

print("Входные слова:")
print(words)
print("Количество слов:", len(words))
```

```python
lemmatizer = WordNetLemmatizer()

noun_lemmas = [lemmatizer.lemmatize(w, pos="n") for w in words]
verb_lemmas = [lemmatizer.lemmatize(w, pos="v") for w in words]

print(f"{'Слово':<12}{'Существительное':<17}{'Глагол'}")
print("-" * 42)
for w, n, v in zip(words, noun_lemmas, verb_lemmas):
    print(f"{w:<12}{n:<17}{v}")
```
