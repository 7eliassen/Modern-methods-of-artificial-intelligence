# Задание 2. Преобразование слов в их базовые формы с помощью стемминга

Используются три стеммера NLTK: `PorterStemmer`, `LancasterStemmer` и `SnowballStemmer`.

```python
from nltk.stem.porter import PorterStemmer
from nltk.stem.lancaster import LancasterStemmer
from nltk.stem.snowball import SnowballStemmer
```

```python
words = ["sing", "sang", "singer", "singing", "calves", "cats", "running", "runs",
         "ran", "happily", "happiness", "generously", "argue", "argued", "university",
         "universal", "fairly", "arrival", "gently", "maximum"]

print("Входные слова:")
print(words)
print("Количество слов:", len(words))
```

```python
porter = PorterStemmer()
lancaster = LancasterStemmer()
snowball = SnowballStemmer("english")

porter_stems = [porter.stem(w) for w in words]
lancaster_stems = [lancaster.stem(w) for w in words]
snowball_stems = [snowball.stem(w) for w in words]

print(f"{'Слово':<12}{'Porter':<12}{'Lancaster':<12}{'Snowball'}")
print("-" * 48)
for w, p, l, s in zip(words, porter_stems, lancaster_stems, snowball_stems):
    print(f"{w:<12}{p:<12}{l:<12}{s}")
```
