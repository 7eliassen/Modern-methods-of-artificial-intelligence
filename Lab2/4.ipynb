# Задание 4. Разбиение текстовых данных на информационные блоки

Чанкинг: разбиение текста на блоки фиксированного размера. Используется корпус Brown из NLTK.

```python
import nltk
from nltk.corpus import brown

# Корпус Brown
nltk.download("brown", quiet=True)
```

```python
N = 1000  # количество первых слов из корпуса Brown (не менее 1000)
M = 20    # количество слов в одном чанке

words = brown.words()[:N]
print("Всего слов:", len(words))
print(" ".join(words[:30]))
```

```python
chunks = [words[i:i + M] for i in range(0, len(words), M)]
print("Количество чанков:", len(chunks))

# Чанков много, выводим первые 20; чанки велики, выводим первые 50 символов
for i, chunk in enumerate(chunks[:20], 1):
    text = " ".join(chunk)
    print(f"{i:2}: {text[:50000]}")
```
